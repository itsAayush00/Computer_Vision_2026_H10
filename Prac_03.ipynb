Importing all the required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

Importing and Reading Image

```python
img = cv2.imread('f2a01e86c920a43c10b65128e272f94f.jpg', cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap='gray')
plt.show()
img.shape
```

Defining the MEAN Function

```python
def mean_filter(img, k=3, s=1):
    pad = k // 2
    padded = np.pad(img, pad, mode='reflect')

    h,w = img.shape
    out_h = (h - k) // s + 1
    out_w = (w - k) // s + 1

    og = np.zeros((out_h, out_w), dtype=np.uint8)

    for i in range(out_h):
        for j in range(out_w):
            window = padded[i*s:i*s+k, j*s:j*s+k]
            og[i,j] = np.mean(window)
    return og
```

Calling the MEAN Function

```python
mean_filtered = mean_filter(img, 3, 1)
plt.imshow(mean_filtered, cmap='gray')
```

Defining the MEDIAN Filter

```python
def median_filter(img, k=3, s=1):
    pad = k // 2
    padded = np.pad(img, pad, mode='reflect')

    h, w = img.shape
    out_h = (h - k) // s + 1
    out_w = (w - k) // s + 1

    out = np.zeros((out_h, out_w), dtype=np.uint8)

    for i in range(out_h):
        for j in range(out_w):
            window = padded[i*s:i*s+k, j*s:j*s+k]
            out[i, j] = np.median(window)

    return out
```

Calling the MEDIAN Filter

```python
median_filtered = median_filter(img, 3, 1)
plt.imshow(median_filtered, cmap='gray')
plt.show()
```
