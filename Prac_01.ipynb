Import the libraries needed for loading, displaying, and processing images.

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

Load the image from the working directory and display it in a window.

```python
img = cv2.imread('f2a01e86c920a43c10b65128e272f94f.jpg')
cv2.imshow('image', img)
cv2.waitKey(0)
```

Check the dimensions of the image array to understand its shape.

```python
img.shape
```

Display the image data type to confirm the array format.

```python
img.dtype
```

Inspect the image data type to understand the pixel value range.

```python
crop = img[300:980, 100:759]
cv2.imshow('crop', crop)
cv2.waitKey(0)
```

Crop a rectangular region from the image to focus on a specific area.

```python
resize = cv2.resize(img, (300, 500))
cv2.imshow('resize', resize)
cv2.waitKey(0)
```

Resize the image to a smaller dimension for easier viewing.

```python
img = cv2.imread('f2a01e86c920a43c10b65128e272f94f.jpg')
img[120, 100] = [0, 0, 0]

cv2.imshow('image', img)
cv2.waitKey(0)
```

Modify a single pixel to see how image arrays are edited in OpenCV.

```python
img_grey = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imshow('image', img_grey)
cv2.waitKey(0)
```

Convert the image to grayscale to demonstrate color-space transformation.

```python
img = cv2.imread('f2a01e86c920a43c10b65128e272f94f.jpg')

img_bgr = img
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.subplot(1, 2, 1)
plt.imshow(img_bgr)

plt.subplot(1, 2, 2)
plt.imshow(img_rgb)

plt.show()
```

Compare the same image displayed as BGR and as RGB to show the channel-order difference between OpenCV and Matplotlib.

```python
red_channel = img[:, :, 2]
plt.imshow(red_channel, cmap='gray')
plt.show()
```

Homework: New Image Operations

```python
#Flip the image horizontally
flipped = cv2.flip(img, 1)
cv2.imshow('flipped', flipped)
cv2.waitKey(0)
```

```python
#Rotate the image by 90 degrees
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
cv2.imshow('rotated', rotated)
cv2.waitKey(0)
```

```python
#Create a blurred version using a Gaussian filter
blurred = cv2.GaussianBlur(img, (15, 15), 0)
cv2.imshow('blurred', blurred)
cv2.waitKey(0)
```
